```python
import pandas as pd

# Load the dataset
dataset_url = 'https://raw.githubusercontent.com/salemprakash/EDA/main/Data/TravelMode.csv'
df = pd.read_csv(dataset_url)

# -------------------------------
# Step 1: Check for missing values
# -------------------------------
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

# -------------------------------
# Step 2: Handle missing values
# -------------------------------

# Separate numerical and categorical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

# Fill missing numerical values with the median
for col in num_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].median(), inplace=True)

# Fill missing categorical values with the mode
for col in cat_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].mode()[0], inplace=True)

# If any rows still contain missing values, remove them
df.dropna(inplace=True)

# -------------------------------
# Step 3: Verify
# -------------------------------
print("\nMissing values after handling:")
print(df.isnull().sum())

print("\nTotal missing values after handling:", df.isnull().sum().sum())

# Display first few rows
print("\nUpdated Dataset:")
print(df.head())
```

```python
import pandas as pd

# 1 Display basic information

print("Dataset Shape:", df.shape)
print("\nData Types:")
print(df.dtypes)


# 2 Check for missing values

print("\nMissing Values:")
print(df.isnull().sum())

# Fill missing numerical values with median

num_cols = df.select_dtypes(include=['int64', 'float64']).columns
for col in num_cols:
    df[col].fillna(df[col].median(), inplace=True)

# Fill missing categorical values with mode
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col].fillna(df[col].mode()[0], inplace=True)


# 3 Check for duplicate rows

duplicates = df.duplicated().sum()
print("\nDuplicate Rows:", duplicates)

# Remove duplicate rows
df.drop_duplicates(inplace=True)


# 4 Remove leading/trailing spaces

for col in cat_cols:
    df[col] = df[col].str.strip()

# 5 Verify cleaned dataset

print("\nFinal Dataset Shape:", df.shape)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Rows After Cleaning:", df.duplicated().sum())

print("\nFirst 5 Rows of Cleaned Dataset:")
print(df.head())
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler


# 1 Label Encoding

le = LabelEncoder()

# Encode all categorical columns
cat_cols = df.select_dtypes(include=['object']).columns

for col in cat_cols:
    df[col] = le.fit_transform(df[col])

print("After Label Encoding:")
print(df.head())


# 2 Feature Scaling

scaler = StandardScaler()

# Select numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

# Scale numerical columns
df[num_cols] = scaler.fit_transform(df[num_cols])

print("\nAfter Feature Scaling:")
print(df.head())


# 3. Verify transformed data

print("\nDataset Information:")
print(df.info())

print("\nStatistical Summary:")
print(df.describe())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Display columns
print(df.columns)

#  Pair Plot

# Select numerical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

sns.pairplot(df[num_cols])
plt.suptitle("Pair Plot of TravelMode Dataset Variables", y=1.02)
plt.show()

#Correlation Heatmap
plt.figure(figsize=(10,6))

correlation = df.corr(numeric_only=True)

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap of TravelMode Dataset")
plt.show()

#Scatter Plot with Multiple Variables
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x='travel',
    y='gcost',
    hue='choice',
    size='wait',
    sizes=(40,200)
)

plt.title("Travel Cost vs Travel Time by Choice and Waiting Time")
plt.xlabel("Travel Time")
plt.ylabel("Cost")
plt.show()

#Optional: 3D Scatter Plot
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(8,6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    df['travel'],
    df['gcost'],
    df['wait']
)

ax.set_xlabel("Travel Time")
ax.set_ylabel("Cost")
ax.set_zlabel("Waiting Time")

plt.title("3D Visualization of Travel Variables")
plt.show()
```
