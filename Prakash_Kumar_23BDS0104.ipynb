```python
import pandas as pd

# Load the dataset
dataset_url = 'https://raw.githubusercontent.com/salemprakash/EDA/main/Data/TravelMode.csv'
df = pd.read_csv(dataset_url)

# -------------------------------
# Step 1: Check for missing values
# -------------------------------
print("Missing values in each column:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())

# -------------------------------
# Step 2: Handle missing values
# -------------------------------

# Separate numerical and categorical columns
num_cols = df.select_dtypes(include=['int64', 'float64']).columns
cat_cols = df.select_dtypes(include=['object']).columns

# Fill missing numerical values with the median
for col in num_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].median(), inplace=True)

# Fill missing categorical values with the mode
for col in cat_cols:
    if df[col].isnull().sum() > 0:
        df[col].fillna(df[col].mode()[0], inplace=True)

# If any rows still contain missing values, remove them
df.dropna(inplace=True)

# -------------------------------
# Step 3: Verify
# -------------------------------
print("\nMissing values after handling:")
print(df.isnull().sum())

print("\nTotal missing values after handling:", df.isnull().sum().sum())

# Display first few rows
print("\nUpdated Dataset:")
print(df.head())
```

K-MEANS CLUSTERING

```python
# Prepare Data for Clustering
cluster_columns = [
    'wait',
    'vcost',
    'travel',
    'gcost',
    'income',
    'size'
]

X = df[cluster_columns].copy()

display(X.head())

# Check Missing Values Before Clustering
print(X.isnull().sum())

# Standardization
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(
    X_scaled,
    columns=cluster_columns
)

display(X_scaled_df.head())

# Elbow Method
wcss = []

K = range(1, 11)

for k in K:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    kmeans.fit(X_scaled)

    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))

plt.plot(
    K,
    wcss,
    marker='o'
)

plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')

plt.title('Elbow Method for Optimal Number of Clusters')

plt.xticks(K)

plt.show()

# Silhouette Score
silhouette_scores = {}

for k in range(2, 11):

    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X_scaled)

    score = silhouette_score(
        X_scaled,
        labels
    )

    silhouette_scores[k] = score

silhouette_df = pd.DataFrame(
    list(silhouette_scores.items()),
    columns=['Number of Clusters', 'Silhouette Score']
)

display(silhouette_df)

# Plot Silhouette Scores
plt.figure(figsize=(8, 5))

plt.plot(
    silhouette_df['Number of Clusters'],
    silhouette_df['Silhouette Score'],
    marker='o'
)

plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')

plt.title('Silhouette Score for Different K Values')

plt.show()

# Find Best K Automatically
best_k = max(
    silhouette_scores,
    key=silhouette_scores.get
)

print("Best Number of Clusters:", best_k)
print(
    "Best Silhouette Score:",
    silhouette_scores[best_k]
)

# Build Final K-Means Model
kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

df['KMeans_Cluster'] = kmeans.fit_predict(X_scaled)

print("K-Means clustering completed.")

# Display Cluster Assignment
display(
    df[
        [
            'mode',
            'choice',
            'travel',
            'gcost',
            'income',
            'KMeans_Cluster'
        ]
    ].head(20)
)

# Number of Observations in Each Cluster
print(
    df['KMeans_Cluster']
    .value_counts()
    .sort_index()
)

# K-Means Cluster Statistics
kmeans_summary = df.groupby(
    'KMeans_Cluster'
)[cluster_columns].mean()

display(kmeans_summary.round(2))

# K-Means Visualization — Travel vs Cost
plt.figure(figsize=(9, 6))

sns.scatterplot(
    data=df,
    x='travel',
    y='gcost',
    hue='KMeans_Cluster',
    palette='viridis',
    s=70
)

plt.title('K-Means Clusters: Travel Time vs Generalized Cost')

plt.show()

# K-Means 3D Visualization
fig = plt.figure(figsize=(10, 7))

ax = fig.add_subplot(
    111,
    projection='3d'
)

scatter = ax.scatter(
    df['travel'],
    df['gcost'],
    df['income'],
    c=df['KMeans_Cluster'],
    cmap='viridis',
    alpha=0.7
)

ax.set_xlabel('Travel Time')
ax.set_ylabel('Generalized Cost')
ax.set_zlabel('Income')

ax.set_title('K-Means Clustering in 3D')

plt.colorbar(
    scatter,
    label='Cluster'
)

plt.show()

```

COMPARE K-MEANS AND HIERARCHICAL

```python
# Re-create variables from previous cells for K-Means and Hierarchical Clustering
# This ensures the necessary variables (X_scaled, best_k, KMeans_Cluster, Hierarchical_Cluster) are defined.

# Prepare Data for Clustering (from MjUBF-Y6sAnE)
cluster_columns = [
    'wait',
    'vcost',
    'travel',
    'gcost',
    'income',
    'size'
]

X = df[cluster_columns].copy()

# Standardization
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Elbow Method to find best_k (from MjUBF-Y6sAnE)
wcss = []
K = range(1, 11)
for k in K:
    kmeans_model = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans_model.fit(X_scaled)
    wcss.append(kmeans_model.inertia_)

silhouette_scores = {}
for k in range(2, 11):
    model = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = model.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    silhouette_scores[k] = score

best_k = max(silhouette_scores, key=silhouette_scores.get)

# Build Final K-Means Model and assign clusters (from MjUBF-Y6sAnE)
kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)
df['KMeans_Cluster'] = kmeans.fit_predict(X_scaled)

# Agglomerative Hierarchical Clustering and assign clusters (from mmqS6zCzt15e)
hierarchical_model = AgglomerativeClustering(
    n_clusters=best_k,
    linkage='ward'
)
df['Hierarchical_Cluster'] = hierarchical_model.fit_predict(
    X_scaled
)

# Original code from DupsnVv-upyg starts here
# Silhouette Score Comparison
kmeans_silhouette = silhouette_score(
    X_scaled,
    df['KMeans_Cluster']
)

hierarchical_silhouette = silhouette_score(
    X_scaled,
    df['Hierarchical_Cluster']
)

print(
    "K-Means Silhouette Score:",
    round(kmeans_silhouette, 4)
)

print(
    "Hierarchical Silhouette Score:",
    round(hierarchical_silhouette, 4)
)

# Comparison Table
comparison = pd.DataFrame({
    'Method': [
        'K-Means',
        'Hierarchical'
    ],

    'Number of Clusters': [
        best_k,
        best_k
    ],

    'Silhouette Score': [
        kmeans_silhouette,
        hierarchical_silhouette
    ]
})

display(comparison.round(4))

# Compare Cluster Assignments
display(
    df[
        [
            'mode',
            'choice',
            'travel',
            'gcost',
            'income',
            'KMeans_Cluster',
            'Hierarchical_Cluster'
        ]
    ].head(30)
)

# Relationship Between K-Means and Hierarchical Clusters
cluster_comparison = pd.crosstab(
    df['KMeans_Cluster'],
    df['Hierarchical_Cluster']
)

display(cluster_comparison)
```
